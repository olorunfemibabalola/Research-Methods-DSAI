# What is Data Augmentation?
Data Augmentation is a technique used in machine learning and deep learning to increase the diversity and size of a training dataset without actually collecting new data. It involves applying a variety of transformations or modifications to existing data samples, such as images, text, or audio, to create new, slightly altered versions.

# Key Purpose
- To improve model generalization by exposing it to varied examples.
- To reduce overfitting, especially when the dataset is small or imbalanced.
- To make models more robust to real-world variations (e.g., rotation, lighting, noise).

You can observe that this is just a single image file. However, training a deep learning model typically requires a large amount of data. By applying data augmentation techniques, you can generate multiple variations of the same image under different visual conditions—such as changes in orientation, lighting, or scale. This approach effectively expands a small dataset into a much larger and more diverse one, improving the model’s ability to generalize and perform well on unseen data.

```python
#write your code
from torchvision import transforms
from PIL import Image
import matplotlib.pyplot as plt
```

```python
#Defining data augmentation transfomations
transform = transforms.Compose([
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(20),
    transforms.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.2),
    transforms.RandomResizedCrop(size=(224,224), scale=(0.9, 1.1)),
    transforms.ToTensor()
    ])

```

```python
#Loading the example image
img = Image.open('/content/IMG_1549.jpeg')
```

```python
#Generation 20 augmented images and visualize in 4x5 grid

plt.figure(figsize=(12, 10))
for i in range(20):
    augmented_img = transform(img)
    plt.subplot(4, 5, i + 1)
    plt.imshow(augmented_img.permute(1,2,0))
    plt.axis('off')
    plt.title(f'Augmented {i+1}', fontsize = 8)

plt.tight_layout()
plt.show()

```

From the output, you can see, data augmentation has helped generate multiple diverse images from a single original image file.
